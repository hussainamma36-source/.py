```python
#encapsulation:(hide the method and variable) refers to the bundling(hiding) of attributes and methods inside a class.
class computer:
    
    def __init__(self):
        self._maxprice=900 #private attribute_,__
        
    def sell(self):
        print("selling price:{}".format(self._maxprice))
        
    def setmaxprice(self,price):
        self._maxprice=price

c1=computer() # object c1
c1.sell()

#change the price
c1._maxprice=1000
c1.sell()

#using sectter
c1.setmaxprice(2000)
c1.sell()
```

```python
class car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("drive")

class boat:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("sail")

class aeroplane:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("fly")

c1=car("ford","tata")
b1=boat("ibiza","touring")
a1=aeroplane("indigo","economic")
for x in(c1,b1,a1):
    x.move()
    
```

```python
#polymophrism:many forms and in programming its refers to methods and function and operators with the same name that can be executed on many objects or classes.
class car:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("drive")
        
class boat:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("sail")
        
class aeroplane:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def  move(self):
        print("fly")
        
class bike:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("ride")

c1=car("ford","tata")
b1=boat("ibiza","touring")
a2=aeroplane("indigo","economic")
b2=bike("honda","yamaha")
for x in (c1,b1,a2,b2):
    x.move()
```

```python
#inheritence class polymorphism 
class vehicle:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model
    def move(self):
        print("move")

class car(vehicle):
    pass
class boat(vehicle):
    def move(self):
        print("sail")

class plane(vehicle):
    def move(self):
        print("fly")

c1=car("ford","tata")
b1=boat("ibiza","touring")
p1=plane("indigo","economic")
for x in(c1,b1,p1):
    print(x.brand)
    print(x.model)
    x.move()
    
```

```python
#data abstraction
from abc import ABC,abstractmethod

class car(ABC):
    def __init__(self,brand,model,year):
        self.brand=brand
        self.model=model
        self.year=year
    @abstractmethod
    def printdetails(self):
        pass
    def acclerate(self):
        print("speed")
    def break_applied(self):
        print ("car stopped")

class hatchback(car):
    def printdetails(self):
        print("brand:",self.brand)
        print("model:",self.model)
        print("year:",self.year)
    def sunroof(self):
        print("not having this feature")

class suv(car):
    def printdetails(self):
        print("brand:",self.brand)
        print("model:",self.model)
        print("year:",self.year)
    def sunroof(self):
        print("available")

car1=hatchback("maruthi","altu","2022")
car1.printdetails()
car1.acclerate()
car1.sunroof()    
```
